```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-4,4, 14) # Inicio, Fin, Cantidad de datos
y = np.exp(x)

plt.plot(x,y, color = "cornflowerblue", marker ="*", linestyle = "-", linewidth = 1)
plt.title("$f(x)=e^2$")
plt.ylabel("$y$")
plt.xlabel("$x$")
plt.grid() #cuadricula
plt.show()
```

```python
datos = [10,15,10,15,20,24,24,24,30,45,45,30,50,51,55,23,59,60,50,23,24,80]
plt.hist(datos, bins = 8, color = "darkseagreen", edgecolor ="darkred") #bins, cantidad de intervalos
plt.title("Histograma de edades")
plt.xlabel("Edades")
plt.ylabel("Frecuencia")
plt.show()
```

```python
categorias = ["Manzanas" , "Peras", "Duraznos", "Fresas"]
valores = [50, 20, 15, 150]
plt.pie(valores, labels=categorias, autopct='%1.1f%%', startangle=0) #autopct - porcentaje
plt.legend(loc="upper left", bbox_to_anchor=(1, 0.5))
plt.title('Gráfico frutas')
plt.show()
```
